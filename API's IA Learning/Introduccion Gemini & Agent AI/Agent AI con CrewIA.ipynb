```python
from google import genai
import litellm
from crewai import Agent, Task, Crew, LLM
from getpass import getpass
GEMINI_API_KEY = getpass("Enter Gemini API Key: ")
```

```python
#Definir los agentes
profesor = Agent(
    role="Profesor de Inglés",
    goal="Explicar conceptos de gramática y vocabulario en inglés de manera clara y efectiva.",
    backstory="Eres un profesor con experiencia en eseñar inglés a estudiantes de todos los niveles.",
    verbose=True,
    allow_delegation=False,
    llm=LLM(model='gemini/gemini-2.5-flash', api_key=GEMINI_API_KEY),
)

evaluador = Agent(
    role="Evaluador de Inglés",
    goal="Diseñar preguntas de evaluacion y revisar las respuesta del estudiante.",
    backstory="Eres un experto en pruebas de inglés y puedes dar retroalimentación detallada.",
    verbose=True,
    allow_delegation=False,
    llm=LLM(model='gemini/gemini-2.5-flash', api_key=GEMINI_API_KEY),
)
```

```python
#Definir Tarea de los agentes
explicacion_ingles=Task(
    description="Explica la diferencia entre los tiempos verbales 'Present Perfect' y 'Past Simple' con ejemplos.",
    expected_output="Una explicación clara y ejemplos concretos.",
    agent=profesor,
)

evaluacion_ingles = Task(
    description="Crea un test de 3 preguntas basado en la explicacion dad por el profesor y evalua respuestas.",
    expected_output="Un test de 3 preguntas con sus respuestas correctas, relacionado con la diferencia entre 'Present Perfect' y 'Past Simple.",
    agent=evaluador,
)
```

```python
#Definir el Crew
crew = Crew(
    agents=[profesor, evaluador],
    tasks=[explicacion_ingles, evaluacion_ingles],
    verbose=True,
)
```

```python
#Comenzar el crew
resultado = crew.kickoff()
print("Desde aqui es el resultado de los agentes!")
print(resultado)
##Podemos generar y conecta a bd vectoriales
```
