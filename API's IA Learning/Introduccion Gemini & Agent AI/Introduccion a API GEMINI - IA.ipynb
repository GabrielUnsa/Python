## Usando Templates

```python
##Prompt_Template
prompt_template="""
Eres un experto programador de python, escribes tu codigo limpio y lo comentas correctamente segun el Standart.
A continuación, {pregunta}
"""

pregunta="escribe un modulo para que la salida de la consola que este en Markdonw sea legible"

# escribiendo prompt completo
prompt = prompt_template.format(pregunta=pregunta);

print(prompt)
```

```python
client = genai.Client(api_key=GEMINI_API_KEY)

response = client.models.generate_content(
    model='gemini-2.5-flash',
    contents=[prompt],
    config=types.GenerateContentConfig(
        temperature=0.5,
        thinking_config=types.ThinkingConfig(thinking_budget=0) # Disables thinking
    )

);
print(response.text)
```

## R.A.G. - Retrieval Augmented Generation

```python
#Uso de base de datos externas, la ia solo lee el contexto que le pasamos no hace pregunta a exterior
documentos = {
    "IA": "Datos sobre IA",
    "RAG": "Que es RAG?",
    "Machine Learning": "Aprendizaje automatico"
};
```

```python
from getpass import getpass
GEMINI_API_KEY = getpass('Enter API Key: ')
```

```python
from google import genai
client = genai.Client(api_key=GEMINI_API_KEY)
```

```python
def recuperar_contexto(pregunta):
    for tema, contenido in documentos.items():
        if tema.lower() in pregunta.lower():
            return contenido
    return "No se encontraron documentos relevantes."
```

```python
def generar_respuesta(pregunta):
  contexto = recuperar_contexto(pregunta)
  prompt = f"""
    usa el siguiente contexto para responde la pregunta de manera clara y concisa.
    Contexto:{contexto} \n \n Pregunta: {pregunta} """
  response = client.models.generate_content(
     model='gemini-2.5-flash',
     contents=[prompt]
   );
  return response.text;
```

```python
#Test
pregunta_usuario = "Que es Machine Learning?"
respuesta = generar_respuesta(pregunta_usuario);
print("Respuesta: ", respuesta)
```

```python
# Test que no quiero que sepan el contexto
pregunta_usuario = "Que hacer en malaga?"
respuesta = generar_respuesta(pregunta_usuario);
print("Respuesta: ", respuesta)
```

## Agentes

```python
#Agente GENAI
from google import genai

from getpass import getpass
GEMINI_API_KEY = getpass('API KEY: ');

client = genai.Client(api_key = GEMINI_API_KEY);
```

```python
def generar_ejercicio(tema="vocabulario", nivel="intermedio"):
  prompt = f"""
    Eres un tutor de inglés. Genera un ejercicio de {tema} para un estudiante de nivel {nivel}.
    Usa preguntas interactivas y al final proporciona la respuesta correcta para que el estudiante pueda verificar.
  """
  response = client.models.generate_content(
      model='gemini-2.5-flash', contents=[prompt]
  );
  return response.text
```

```python
print("Ejercicio de vocabulario: ")
print(generar_ejercicio(tema="vocabulario", nivel="principiante"))
```

```python
print("Ejercicio de Gramatica: ")
print(generar_ejercicio(tema="gramatica", nivel="intermedio"))
```
